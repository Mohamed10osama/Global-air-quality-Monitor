```python
# Setting the data base #
conn = sql.connect("air_quality.db")
cursor = conn.cursor()
cursor.execute("""
    CREATE TABLE IF NOT EXISTS air_quality (
        city TEXT,
        date TEXT,
        pm25 REAL
    )
""")
cursor.execute("DELETE FROM air_quality")
conn.commit()
```

```python
#Kobe
for j in data2["results"]:
    cursor.execute(
        "INSERT INTO air_quality VALUES (?, ?, ?)",
        ("Kobe", j["period"]["datetimeFrom"]["local"], j["value"])
    )
#Paris
for j in data_paris["results"]:
    cursor.execute(
        "INSERT INTO air_quality VALUES (?, ?, ?)",
        ("Paris", j["period"]["datetimeFrom"]["local"], j["value"])
    )
#Berlin
for j in data_berlin["results"]:
    cursor.execute(
        "INSERT INTO air_quality VALUES (?, ?, ?)",
        ("Berlin", j["period"]["datetimeFrom"]["local"], j["value"])
    )
conn.commit()
cursor.execute("SELECT city, COUNT(*) FROM air_quality GROUP BY city")
print(cursor.fetchall())
```

```python
cursor.execute("Select date,pm25 FROM air_quality WHERE city = 'Kobe'")
kobe_data=cursor.fetchall()
print(kobe_data[:3])
print("#"*50)
cursor.execute("Select date,pm25 FROM air_quality WHERE city = 'Paris'")
paris_data=cursor.fetchall()
print(paris_data[:3])
print("#"*50)
cursor.execute("Select date,pm25 FROM air_quality WHERE city = 'Berlin'")
berlin_data=cursor.fetchall()
print(berlin_data[:3])
```

```python
kobe_values = [j[1] for j in kobe_data]
paris_values = [j[1] for j in paris_data]
berlin_values = [j[1] for j in berlin_data]
```

```python
#MATPLOTLIB 
plt.plot(kobe_values, label="Kobe")
plt.plot(paris_values, label="Paris")
plt.plot(berlin_values, label="Berlin")
plt.legend()
plt.title("PM2.5 Air Quality Comparison")
plt.ylabel("PM2.5 µg/m³")
plt.show()
```
